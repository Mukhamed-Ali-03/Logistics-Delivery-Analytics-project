```python
# Import Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Cleaned Dataset
df = pd.read_csv("../data/amazon_delivery_cleaned.csv")
```

```python
# Business Overview

print(f"Total Orders: {df['Order_ID'].count()}")
print(f"Average Delivery Time: {df['Delivery_Time'].mean():.2f} minutes")
print(f"Average Agent Rating: {df['Agent_Rating'].mean():.2f}")
print(f"Average Agent Age: {df['Agent_Age'].mean():.2f} years")
```

```python
#delivery time by weather
weather = df.groupby("Weather")["Delivery_Time"].mean().sort_values()

weather
```

```python
weather.plot(kind="bar")

plt.title("Average Delivery Time by Weather")
plt.xlabel("Weather")
plt.ylabel("Delivery Time (Minutes)")
plt.xticks(rotation=45)

plt.show()
```

```python
### Insight 
# cloudy and fogy weather generally increase delivery time compared to other weather conditions.
```

```python
#delivery time by traffic 
traffic = df.groupby("Traffic")["Delivery_Time"].mean().sort_values()

traffic
```

```python
traffic.plot(kind="bar")

plt.title("Average Delivery Time by Traffic")
plt.xlabel("Traffic")
plt.ylabel("Delivery Time (Minutes)")

plt.show()
```

```python
#jam leads to longer delivery times.
```

```python
#delivery time by vehicle
vehicle = df.groupby("Vehicle")["Delivery_Time"].mean().sort_values()

vehicle
```

```python
vehicle.plot(kind="bar")

plt.title("Average Delivery Time by Vehicle")
plt.xlabel("Vehicle")
plt.ylabel("Delivery Time (Minutes)")

plt.show()
```

```python
#motorcycle deliveries takes longer time compared to other vehicles.
```

```python
# delivery time by area 
area = df.groupby("Area")["Delivery_Time"].mean().sort_values()

area
```

```python
#delivery time by area 
area = df.groupby("Area")["Delivery_Time"].mean().sort_values()

area.plot(kind="bar")

plt.title("Average Delivery Time by Area")
plt.ylabel("Delivery Time")

plt.show()
```

```python
#deliveries in semi-urban takes longer time compared to other areas.
```

```python
#delivery time by category
category = df.groupby("Category")["Delivery_Time"].mean().sort_values()

category
```

```python

category.plot(kind="bar")

plt.title("Average Delivery Time by Category")
plt.xticks(rotation=45)

plt.show()
```

```python
#groceries deliveries are generally faster compared to other categories.
```

```python
#distribution of delivery time
plt.hist(df["Delivery_Time"], bins=10)

plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time")
plt.ylabel("Orders")

plt.show()
```

```python
#Most deliveries are completed within a moderate time range.
```

```python
#agent age vs delivery time
plt.scatter(df["Agent_Age"], df["Delivery_Time"])

plt.title("Agent Age vs Delivery Time")
plt.xlabel("Agent Age")
plt.ylabel("Delivery Time")

plt.show()
```

```python
#most deliveries were made by the agents with age between 20 to 40 years.
```

```python
#correlation matrix
numeric_df = df.select_dtypes(include="number")

corr = numeric_df.corr()
plt.figure(figsize=(8,6))

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```
