```python
# Import Required Libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
df = pd.read_csv("../data/amazon_delivery.csv")
```

```python
# Missing Values
missing_values = df.isnull().sum()

missing_values
missing_values[missing_values > 0]
```

```python
# Duplicate Records

duplicates = df.duplicated().sum()

print(f"Duplicate Rows: {duplicates}")
```

```python
# Data Types

df.dtypes
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Order_Date"].dtype
```

```python
print(df.info())
```

```python
# Save Cleaned Dataset

df.to_csv("../data/amazon_delivery_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```

```python
## Conclusion
#- Checked missing values.
#- Checked duplicate records.
#- Converted the Order_Date column to datetime format.
#- Verified data types.
#- Saved the cleaned dataset for Exploratory Data Analysis.
```

```python
# Remove leading/trailing spaces
df["Weather"] = df["Weather"].str.strip()

# Replace empty strings with NaN
df["Weather"] = df["Weather"].replace("", pd.NA)

# Remove rows where Weather is missing
df = df.dropna(subset=["Weather"])
```

```python
# Save Cleaned Dataset

df.to_csv("../data/amazon_delivery_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
